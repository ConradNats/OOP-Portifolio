```python
#IF STATEMENTS
score = 80
if score > 80:
    grade = "A"
elif score > 70:
    grade = "B"
elif score > 60:
    grade = "C"
elif score > 50:
    grade = "D"
elif score > 40:
    grade = "E"
elif score > 30:
    grade = "F"
    
print(f"grade:{grade}")
```

```python
#FOR LOOPS
fruits = ["apples", "bananas", "oranges"]
for fruit in fruits:
    print(fruits)
```

```python
rows = 5

for i in range(1, rows + 1):
    for space in range(rows - i):
        print(" ", end="")
    for star in range(2 * i - 1):
        print("*", end="")
    print()
```
